```python
import numpy as np

atac = np.load("../data/atac_matrix.npy")
hic = np.load("../data/hic_matrix.npy")
times = np.load("../data/timepoints.npy")

print("ATAC shape:", atac.shape)
print("HiC shape:", hic.shape)
print("Timepoints:", times)

print("\nATAC small preview:")
print(atac[:2, :10])  

print("\nHiC small preview:")
print(hic[:2, :10])   
```

```python
print("ATAC stats: min", atac.min(), "max", atac.max(), "mean", atac.mean())
print("HiC stats: min", hic.min(), "max", hic.max(), "mean", hic.mean())
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,4))
for i in range(atac.shape[0]):
    plt.hist(atac[i], bins=100, alpha=0.5, label=f"ATAC {times[i]}h")
plt.legend()
plt.title("ATAC signal distribution per timepoint")
plt.xlabel("Signal intensity")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,4))
for i in range(hic.shape[0]):
    plt.hist(np.random.choice(hic[i], 50000), bins=100, alpha=0.5, label=f"HiC {times[i]}h")
plt.legend()
plt.title("HiC interaction intensity distribution")
plt.xlabel("Interaction value")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```

```python
import pandas as pd

df = pd.DataFrame({
    "time": times,
    "mean_signal": atac.mean(axis=1),
    "std_signal": atac.std(axis=1)
})
print(df)
```
